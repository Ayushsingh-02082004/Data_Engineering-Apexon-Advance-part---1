```python
#ARithmatic operation on pandas series

series1 = pd.Series([2, 4, 6, 8, 10])
series2 = pd.Series([1, 3, 5, 7, 9])

print("Addition\n",[ series1 + series2])
print("Addition\n",[ series1 - series2])
print("Addition\n",[ series1 * series2])
print("Addition\n",[ series1 / series2])
```

```python
#Element-wise  Power of Array Values

# Creating the base Series
ds1 = pd.Series([0, 1, 2, 3, 4, 5, 6])

# Creating the exponent Series (all 3s to match your expected output)
ds2 = pd.Series([3, 3, 3, 3, 3, 3, 3])

result = ds1.pow(ds2)

print(ds1)
print(ds2)
print(result)
```
