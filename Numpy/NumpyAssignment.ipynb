```python
#to find the number of elements of an array, length of one
import numpy as np

arr = np.array([10,20,30,40])
size = len(arr)
itemsize = arr.itemsize
print(itemsize)
print(size)
print("Total bytes = " , size*itemsize)
```

```python
# Find Common values between two arrays.
arr = np.array([0,10,20,30,40,60,])
prr = np.array([10,30,40,80,60])

commonvalues = np.intersect1d(arr , prr)
print(arr)
print(prr)
print(commonvalues)
```

```python
#present in one set difference
arr = np.array([0,10,20,30,40,60,90,75,99])
prr = np.array([10,30,40,80,60])

ans = np.setdiff1d(arr,prr)
print(ans)
```

```python
#present in one not both 

arr = np.array([0,10,20,30,40,60,90,75,99])
prr = np.array([10,30,40,80,60])

ans = np.setxor1d(arr,prr)
print(ans)
```

```python
#Compare Arrays

import numpy as np

# Define arrays
a = np.array([1, 2])
b = np.array([4, 5])

# Comparisons
print("a > b")
print(a > b)

print("a >= b")
print(a >= b)

print("a < b")
print(a < b)

print("a <= b")
print(a <= b)
```
