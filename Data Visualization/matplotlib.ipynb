```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("Xaxis")
plt.ylabel("Yaxis")
```

```python
#subplots in matplotlib
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2 , x**2)
```

```python
plt.subplot(6,6,1)
```

```python
# Object Oriented Way
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.8 , 0.1 , .8 ,.8])
axis1.plot(x,y)
axis2 = fig.add_axes([.3 , .4 , .3 , .3])
axis2.plot(x,y)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
plt.subplot(1,2,1)
sns.distplot(df['total_bill'],bins=20 , kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

JointPlot

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'scatter' )
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'kde' )
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'hex' )
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'reg' )
```

```python
sns.pairplot(df)
```
